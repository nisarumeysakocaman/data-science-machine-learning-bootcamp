```python
#indirim_orani ve musteri_puani sütunlarındaki eksik değerleri medyan ile doldurun.
df_medyan=df.copy()
sutunlar=["indirim_orani","musteri_puani"]
for sutun in sutunlar:
  df_medyan[sutun]=df_medyan[sutun].fillna(df_medyan[sutun].median())
df_medyan[sutunlar].isnull().sum()
```

```python
#odeme_turu ve musteri_tipi sütunlarındaki eksik değerleri mod ile doldurun. Daha sonra bu sütunlarda eksik değer kalıp kalmadığını kontrol
df_mod=df_medyan.copy()
sutunlar2=["odeme_turu","musteri_tipi"]
for sutun in sutunlar2:
  df_mod[sutun]=df_mod[sutun].fillna(df_mod[sutun].mode()[0])
df_mod[sutunlar2].isnull().sum()
```

```python
 #İstanbul, istanbul, Istanbul gibi aynı anlama gelen farklı yazımları tek bir standart biçime dönüştürün.
def metin_normalize_et(series):
  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇","i",regex=False)
      .str.replace("ş","s",regex=False)
      .str.replace("ı","i",regex=False)
      .str.replace("ç","c",regex=False)
      .str.replace("ö","o",regex=False)
      .str.replace("ğ","g",regex=False)
      .str.replace("ü","u",regex=False)
      .str.replace("ş","s",regex=False)

  )
```

```python
#birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın.
#Sonra aykırı değer içeren kayıt sayısını yazdırın.
df=pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayıtlar_düzenlendi.csv")
df["birim_fiyat"].describe()
```

```python
"""
birim fiyat
IQR= Q3(75%) - Q1(25%)
   = 824 - 198 = 626
Üst Sınır= Q3 + k * IQR
Alt Sınır= Q1 - k* IQR
  k = 1.5 olsun
  Üst sınır=  824 + 1.5*656 = 939
  Alt sınır= 198 - 1.5*656 = -741
Outlier: üst sınır ve alt sınırın dışında kalan örnekler
"""
```

```python
#IQR ile aykırı değer sınırlarını hesaplama

# Çeyrekler ve IQR hesabı
Q1 = df["birim_fiyat"].quantile(0.25)
Q3 = df["birim_fiyat"].quantile(0.75)
IQR = Q3 - Q1

# Sınırların belirlenmesi
alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

# Değerleri sözlükte toplama
birim_fiyat_sinirlar = {
    "q1": Q1,
    "q3": Q3,
    "iqr": IQR,
    "alt_sinir": alt_sinir,
    "ust_sinir": ust_sinir
}

print(birim_fiyat_sinirlar)

```

```python
# Toplam aykırı kayıt sayısı
aykiri_sayisi = ((df["birim_fiyat"] < alt_sinir) | (df["birim_fiyat"] > ust_sinir))

print(f"birim_fiyat sütunundaki toplam aykırı kayıt sayısı: {aykiri_sayisi.sum()}")
df.shape[0]
```

Toplam tutar grafiklerinin daha yorumlanabilir olması için toplam tutardaki aykırı değerleri veri setinden çıkarıyoruz.

```python
df["toplam_tutar"].describe()
```

```python
#IQR ile aykırı değer sınırlarını hesaplama

# Çeyrekler ve IQR hesabı
Q1 = df["toplam_tutar"].quantile(0.25)
Q3 = df["toplam_tutar"].quantile(0.75)
IQR = Q3 - Q1

# Sınırların belirlenmesi
alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

# Değerleri sözlükte toplama
toplam_tutar_sinirlar = {
    "q1": Q1,
    "q3": Q3,
    "iqr": IQR,
    "alt_sinir": alt_sinir,
    "ust_sinir": ust_sinir
}

print(toplam_tutar_sinirlar)
```

```python
# Toplam tutar aykırı kayıt sayısı
aykiri_sayisi = ((df["toplam_tutar"] < alt_sinir) | (df["toplam_tutar"] > ust_sinir))

print(f"toplam_tutar sütunundaki toplam aykırı kayıt sayısı: {aykiri_sayisi.sum()}")
df.shape[0]
```

Aykırı değerler genel değerlere kıyasla çok uç değerlerdir ve tablo veya grafik yorumlamalarında hataya sebep olabilir.Bu yüzden toplam tutardaki aykırı değerleri daha stabil bir yorumlama için veri setinden çıkarıyoruz.

```python
# toplam tutar aykırı olan satırları tablodan tamamen çıkarır
df_temiz = df[~aykiri_sayisi].copy().reset_index(drop=True)
df_temiz.shape[0]
```
